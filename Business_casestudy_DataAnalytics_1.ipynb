```python
import numpy as np
import scipy.stats as stats

# Sample data
data = [49, 50, 48, 47, 51, 50, 49, 48, 47, 48]
mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

# 95% Confidence Interval
conf_level = 0.95
z = stats.t.ppf((1 + conf_level) / 2, df=n-1)
margin_of_error = z * (std / np.sqrt(n))

lower = mean - margin_of_error
upper = mean + margin_of_error

print("95% Confidence Interval: [", lower, ",", upper, "]")
```

```python
import numpy as np
from scipy.stats import norm

# Hypothetical values
population_mean = 170     # μ₀
population_std = 10       # σ (known)
sample = np.random.normal(loc=172, scale=10, size=100)  # Sample data

# Step 1: Calculate sample mean and size
sample_mean = np.mean(sample)
n = len(sample)

# Step 2: Calculate z-score
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))

# Step 3: Calculate p-value (two-tailed)
p_value = 2 * (1 - norm.cdf(abs(z_score)))

# Step 4: Print results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-score: {z_score:.2f}")
print(f"P-value: {p_value:.4f}")

# Step 5: Conclusion
alpha = 0.05
if p_value < alpha:
    print("Reject the Null Hypothesis (significant difference).")
else:
    print("Fail to Reject the Null Hypothesis (no significant difference).")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Positive covariance example
x_pos = np.array([1, 2, 3, 4, 5])
y_pos = np.array([2, 4, 6, 8, 10])

# Negative covariance example
x_neg = np.array([1, 2, 3, 4, 5])
y_neg = np.array([10, 8, 6, 4, 2])

# No covariance example
np.random.seed(0)
x_rand = np.random.rand(50)
y_rand = np.random.rand(50)

# Plot
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

# Positive covariance
axs[0].scatter(x_pos, y_pos, color='green')
axs[0].set_title("Positive Covariance")
axs[0].set_xlabel("X")
axs[0].set_ylabel("Y")

# Negative covariance
axs[1].scatter(x_neg, y_neg, color='red')
axs[1].set_title("Negative Covariance")
axs[1].set_xlabel("X")
axs[1].set_ylabel("Y")

# No covariance
axs[2].scatter(x_rand, y_rand, color='gray')
axs[2].set_title("No Covariance")
axs[2].set_xlabel("X")
axs[2].set_ylabel("Y")

plt.suptitle("Covariance Examples: Positive, Negative, None", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data for correlation visualization
# Positive correlation
x_pos = np.array([1, 2, 3, 4, 5])
y_pos = np.array([2, 4, 6, 8, 10])
r_pos = np.corrcoef(x_pos, y_pos)[0, 1]

# Negative correlation
x_neg = np.array([1, 2, 3, 4, 5])
y_neg = np.array([10, 8, 6, 4, 2])
r_neg = np.corrcoef(x_neg, y_neg)[0, 1]

# No correlation (random)
np.random.seed(1)
x_none = np.random.rand(50)
y_none = np.random.rand(50)
r_none = np.corrcoef(x_none, y_none)[0, 1]

# Plot
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

# Positive
axs[0].scatter(x_pos, y_pos, color='green')
axs[0].set_title(f"Positive Correlation\nr = {r_pos:.2f}")
axs[0].set_xlabel("X")
axs[0].set_ylabel("Y")

# Negative
axs[1].scatter(x_neg, y_neg, color='red')
axs[1].set_title(f"Negative Correlation\nr = {r_neg:.2f}")
axs[1].set_xlabel("X")
axs[1].set_ylabel("Y")

# None
axs[2].scatter(x_none, y_none, color='gray')
axs[2].set_title(f"No Correlation\nr = {r_none:.2f}")
axs[2].set_xlabel("X")
axs[2].set_ylabel("Y")

plt.suptitle("Correlation Examples: Positive, Negative, None", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 20
p = 0.5

# X values (number of successes)
x = np.arange(0, n + 1)

# Binomial PMF
y = binom.pmf(x, n, p)

# Plot
plt.figure(figsize=(8, 5))
plt.stem(x, y)
plt.title("Binomial Distribution (n = 20, p = 0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
