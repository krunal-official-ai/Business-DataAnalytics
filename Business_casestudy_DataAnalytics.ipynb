```python
import pandas as pd

# Salaries in ₹
salaries = [30000, 32000, 33000, 31000, 10000000]

# Creating DataFrame
df = pd.DataFrame(salaries, columns=["Salary"])

# Calculating
mean_salary = df["Salary"].mean()
median_salary = df["Salary"].median()

print("Mean Salary: ₹", mean_salary)
print("Median Salary: ₹", median_salary)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Two datasets
data1 = [50, 50, 50, 50, 50]     # No variation
data2 = [30, 40, 50, 60, 70]     # With variation

# Calculate mean and std dev
mean1 = np.mean(data1)
std1 = np.std(data1)
mean2 = np.mean(data2)
std2 = np.std(data2)

# Plotting
plt.figure(figsize=(12, 5))

# Plot 1
plt.subplot(1, 2, 1)
plt.plot(data1, marker='o', linestyle='-', label=f'Std Dev = {std1:.2f}')
plt.axhline(mean1, color='red', linestyle='--', label=f'Mean = {mean1}')
plt.title('Dataset 1: Low Spread')
plt.legend()
plt.grid(True)

# Plot 2
plt.subplot(1, 2, 2)
plt.plot(data2, marker='o', linestyle='-', label=f'Std Dev = {std2:.2f}')
plt.axhline(mean2, color='red', linestyle='--', label=f'Mean = {mean2}')
plt.title('Dataset 2: High Spread')
plt.legend()
plt.grid(True)

plt.suptitle('Visualizing Standard Deviation and Mean', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate normal distribution data
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=1000)

# Plot using seaborn
sns.histplot(data, kde=True, bins=30, color='skyblue', edgecolor='black')
plt.axvline(np.mean(data), color='red', linestyle='--', label='Mean')
plt.title('Gaussian (Normal) Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

data = np.array([45, 55, 60, 63, 65, 68, 70, 72, 75, 80])

min_val = np.min(data)
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)  # median
q3 = np.percentile(data, 75)
max_val = np.max(data)

print(f"Minimum: {min_val}")
print(f"Q1 (25th percentile): {q1}")
print(f"Median (Q2, 50th percentile): {q2}")
print(f"Q3 (75th percentile): {q3}")
print(f"Maximum: {max_val}")
```
